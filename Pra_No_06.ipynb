## *K-Means Clustering using Python Scikit-learn*

```python
# Step 1: Import librarie

import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
# Step 2: Create datase
x =[
    [1,2] ,[2,1],[2,3],
    [8,8], [9,8],[8,9],
    [15,2], [16,3],[15,4]

]
```

```python
# Step 3: Create K-Means model
model = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
    )
```

```python
# Step 4: Apply K-Means clustering

model.fit(x)
```

```python
# Step 5: Display cluster labels

print("Cluster Labels:")
print(model.labels_)
```

```python
# Step 6: Display centroids

print("\nCluster Centroids:")
print(model.cluster_centers_)
```

```python
# Step 7: Visualize clusters
plt.scatter(
    [p[0] for p in x],
    [p[1] for p in x],
    c=model.labels_
)
plt.scatter(
    model.cluster_centers_[:, 0],
    model.cluster_centers_[:, 1],
    marker="X",
    s=500,
    label="Centroids"
)
plt.title("K-Means Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.show()
```
